```python
import pandas as pd 
```

```python
# Creating DataFrames
data = {
    'age': [25, 30, 35, 40],
    'salary': [50000, 60000, 70000, 80000],
    'city': ['NYC', 'SF', 'LA', 'NYC']
}
df = pd.DataFrame(data)

# Reading data (most common in real projects)
# df = pd.read_csv('data.csv')

# Basic exploration
print(df.head())        # First 5 rows
print(df.info())        # Data types, null counts
print(df.describe())    # Statistics for numerical columns

# Selecting data
ages = df['age']                    # Single column
subset = df[['age', 'salary']]      # Multiple columns
filtered = df[df['age'] > 30]       # Filtering rows

# Data cleaning
df_clean = df.dropna()              # Remove rows with missing values
df['age'].fillna(df['age'].mean())  # Fill missing with mean

# Grouping (like SQL GROUP BY)
avg_salary_by_city = df.groupby('city')['salary'].mean()

# Creating new columns
df['salary_in_k'] = df['salary'] / 1000
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
# Scikit-learn workflow (THIS IS THE PATTERN YOU'LL USE ALWAYS)

# Step 1: Prepare data
X = df[['age', 'salary']]  # Features
y = df['city']              # Target

# Step 2: Split data (more on this later)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
print("X_train ",X_train)
print("X_test ",X_test)
print("y_train ",y_train)
print("y_test ",y_test)
# Step 3: Create and train model
model = DecisionTreeClassifier()
print("model ",model)
model.fit(X_train, y_train)  # Learning happens here!

# Step 4: Make predictions
predictions = model.predict(X_test)
print("predictions ",predictions)

# Step 5: Evaluate
accuracy = accuracy_score(y_test, predictions)
print(f"Accuracy: {accuracy}")
```
