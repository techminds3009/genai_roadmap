What is numpy ?
Numpy is python lib used to work with numbers, arrays to perform mathematical operations efficiently. It gives fast numberical computation (faster than python list), powerful multi-dimemsional array(NDarray), tools for linear algebra, stats, random numbers

It is fast because it uses optimized C backend. Operation on array happens at once, much faster then python loops. 

Example: adding 2 list in python (slow loop)
Adding 2 numpy arrays (vectorized, extremely fast)

All ML models uses matrices, vectors, linear algebra, matrix multiplication, dot products. Numpy handles all the efficiently

All major libs uses numpy, like pands, scikiet-leawrn takes numpy arrays as input, pytorch and tensorFlow uses numpy operations

```python
import numpy as np
```

```python
arr1=np.array([1,2,3,4,5])
arr2=np.array([6,7,8,9,10,11])

```

```python
arr1+=1
arr2*=2
print("Array 1 after addition:", arr1)
print("Array 2 after multiplication:", arr2)
```

```python
##since array are of different shapes, broadcasting will fail here a
## is of size (5,), while arr2 is of size (6,)
arr3=arr1+arr2
print("Resultant Array after addition:", arr3)
```

```python
arr3=arr1+arr2[:5]
print("Resultant Array after addition:", arr3)
```

```python
print(arr1.shape)
print(arr2.shape)
```

```python
print(arr1.reshape(5, 1))
print("--------------------------")
print(arr2.reshape(1, 6))
print("--------------------------")
arr3=arr1.reshape(5, 1) + arr2.reshape(1, 6)
#arr3=arr1+arr2
print(arr3)
#print("Element-wise addition:", arr3)
```

array creation
np.linspace() generates evenly spaced numbers between a start and end value.
Why it's useful:
Used in ML & Data Science for plotting graphs
Creating smooth ranges
Creating sample data
Creating equally distributed values

np.eye(3): Identity matrix is like the number 1 in matrices.Multiplying a matrix A × I = A
Used heavily in:
Linear algebra
Inverse matrix checks

```python
zeroes=np.zeros((3,3))               # matrix of zeros
ones=np.ones((2,2))                # matrix of ones
like_range=np.arange(0, 10, 2)           # like range()
linspace_op=np.linspace(0, 1, 5)          # 5 values between 0 and 1
print("Zeros:\n", zeroes)
print("Ones:\n", ones)
print("Arange:\n", like_range)
print("Linspace:\n", linspace_op)
np.eye(3) 
```
